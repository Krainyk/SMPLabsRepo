```python
import pandas as pd
from numpy import dtypes

sweets_prices = {
"Ferrero Rocher": 150,
"Lindt Excellence": 120,
"Raffaello": 130,
"KitKat": 25,
"M&M's": 40,
"Snickers": 20,
"Milka": 35,
"Twix": 22,
"Toblerone": 80,
"Skittles": 30,
"Kinder Chocolate": 45,
"Mars": 20,
"Hershey's": 60,
"Bounty": 22,
"Nestlé Aero": 50
}
```

```python
sweets = pd.Series(sweets_prices, dtype="float64")
print("sweets:")
print(sweets)
```

```python
percent = 10
sweets = sweets * (1 + percent / 100)
print("sweets:")
print(sweets)
```

```python
sweets.loc[(sweets >= 20) & (sweets <= 70)] *= 1.20
print(sweets)
```

```python
print(f"Snickers price: Snickers {sweets['Snickers']}")
print(f"dtype: {sweets.dtypes}")

sweets["Snickers"] += 100
print(f"Updated Snickers price: Snickers {sweets['Snickers']}")
print(f"dtype: {sweets.dtypes}")
```

```python
top_seven = sweets.nlargest(7)
print("top_seven")
print(top_seven)
```
