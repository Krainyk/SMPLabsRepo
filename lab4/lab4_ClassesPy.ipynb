```python
class Character:
    def __init__(self, name, health, level=1):
        self.name = name
        self.health = health
        self.level = level
        self.inventory = []
        self.skills = {}

    def take_damage(self, amount):
        self.health -= amount
        if self.health <= 0:
            self.health = 0
            print(f"{self.name} has been defeated!")

    def heal_hp(self, amount):
        self.health += amount
        if self.health > 100:
            self.health = 100
        print(f"{self.name} healed for {amount} points. Current health: {self.health}")

    def level_up(self):
        self.level += 1
        print(f"{self.name} leveled up to level {self.level}!")

    def add_item(self, item):
        self.inventory.append(item)
        print(f"{item} added to {self.name}'s inventory.")

    def remove_item(self, item):
        if item in self.inventory:
            self.inventory.remove(item)
            print(f"{item} removed from {self.name}'s inventory.")
        else:
            print(f"{item} not found in {self.name}'s inventory.")

    def learn_skill(self, skill_name, skill_level):
        self.skills[skill_name] = skill_level
        print(f"{self.name} learned {skill_name} at level {skill_level}.")

    def show_character_info(self):
        print("=" * 40)
        print(f"Character: {self.name}")
        print(f"Health: {self.health}")
        print(f"Level: {self.level}")
        print(f"Inventory: {self.inventory}")
        print(f"Skills: {self.skills}")
        print("=" * 40)

ch = Character("Gerald", 100, 5)
ch.take_damage(30)
ch.heal_hp(20)
ch.level_up()
ch.add_item("Arondit")
ch.add_item("Potion")
ch.remove_item("Potion")
ch.learn_skill("Aard", 3)
ch.show_character_info()
```

```python
class Sword:
    def __init__(self, name, damage, durability):
        self.name = name
        self.damage = damage
        self.durability = durability

    def attack(self):
        if self.durability > 0:
            self.durability -= 1
            print(f"{self.name} attacks for {self.damage} damage! Durability left: {self.durability}")
        else:
            print(f"{self.name} is broken and cannot be used!")
```

```python
class Witcher(Character):
    def __init__(self, name, health, level=1, school="Wolf"):
        super().__init__(name, health, level)
        self.school = school
        self.monster_kills = 0
        self.sword = Sword("Steel Sword", 15, 100)

    def hunt_monster(self):
        self.monster_kills += 1
        print(f"{self.name} has hunted a monster! Total kills: {self.monster_kills}")

    def show_info(self):
        super().show_character_info()
        print(f"School: {self.school}")
        print(f"Monster Kills: {self.monster_kills}")

    def fight(self):
        print(f"{self.name} is fighting with {self.sword.name}!")
        self.sword.attack()

witcher = Witcher("Geralt", 100, 10, "Wolf")
witcher.hunt_monster()
witcher.hunt_monster()
witcher.fight()
witcher.show_info()
```
