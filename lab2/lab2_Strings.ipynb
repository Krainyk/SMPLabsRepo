```python
def analyze_email(email_address):
    at_index = email_address.find("@")
    # print(at_index)

    login = email_address[:at_index]
    domain = email_address[at_index+1:]

    result = f"(Email,{login},{domain})"
    print(result)
    print("(Email", login, domain + ")", sep="|")

# analyze_email("student.name_group@lpnu.com")
analyze_email("student.nikolas_RI23@lpnu.com")
```

```python
def format_name(first_name, last_name, upper_case=False):
    full_name = f"{first_name} {last_name}"

    if upper_case:
        result = full_name.upper()
    else:
        result = full_name.title()

    return result

print(format_name("nikolas", "krainyk"))
print(format_name("nikolas", "krainyk", upper_case=True))
```

```python
def create_shopping_list(store_name, *items):
    store_headline = f"{store_name}"
    shop_list = ""
    for i, item in enumerate(items, start=1):
        shop_list += f"{i} {item}\n"
        # print(i, item)

    return store_headline, shop_list.strip()

# create_shopping_list("ATB", "milk", "cheese", "bread", "apple")
print(*create_shopping_list("ATB", "milk", "cheese", "bread", "apple"), sep="\n--------\n")
```

```python
def process_tags(tags_string, delimiter=","):
    tags_strip = tags_string.split(delimiter)

    cleaned_tags = []
    for tag in tags_strip:
        clean_tag = tag.strip("!- ")
        cleaned_tags.append(clean_tag)

    final_tags = ' # '.join(cleaned_tags)

    return final_tags

tags_string = " python, data science! , --machine learning"
print(process_tags(tags_string))
# print(process_tags(" python, data science! , --machine learning"))
```

```python
def create_profile(login, password, separator="| ", **user_data):
    profile = ""; profile += f"login{separator}{login}\n"
    passwd = ""; passwd += f"password{separator}{password}\n"

    usr_data = ""

    for key, value in user_data.items():
        usr_data += f"{key}{separator}{value}\n"

    usr_profile = profile + passwd + usr_data

    return usr_profile

print(create_profile("Student", "PI-23", separator=": ", name='Ivan', age=30, occupation='Developer', city='Lviv'))
```
