```python
def analyze_email(email_address):
    at_index = email_address.find("@")
    print(at_index)

    login = email_address[:at_index]; print(login)
    domain = email_address[at_index+1:]; print(domain)

    result = f"(Email,{login},{domain})"; print(result)
    print("(Email", login, domain + ")", sep="|")

# analyze_email("student.name_group@lpnu.com")
analyze_email("student.nikolas_RI23@lpnu.com")
# analyze_email("student.jfewofjeofepfjeofsofjsopfseofesofj_RI23@lpnu.com")
```

```python
def format_name(first_name, last_name, upper_case=False):
    full_name = f"{first_name} {last_name}"

    if upper_case:
        result = full_name.upper()
    else:
        result = full_name.title()

    return result

print(format_name("nikolas", "krainyk"))
print(format_name("nikolas", "krainyk", upper_case=True))
```
