```python
var_start = 1
var_end = 20
total_sum = 0

for i in range(var_start, var_end + 1):
    if i % 2 == 0:
        total_sum += i

print(total_sum)
```

```python
var_start = 1
var_end = 20
var_step = 2
total_mult = 1

for i in range(var_start, var_end + 1, var_step):
    if i % 2 != 0:
        total_mult *= i

print(total_mult)
```

```python
product_name = "Laptop"
price = 501
is_sale = False
is_available = False
category = ""

if not is_available:
    print(f"Товар {product_name} тимчасово відсутній на складі.")
else:
    print(f"Товар {product_name} доступний на складі.")

if price <= 500 or is_sale:
    category = "Budget/Promotional"
elif 501 <= price <= 1000:
    category = "Standard"
else:
    category = "Premium"

print(f"Product: {product_name}, Price: {price}, Category: {category}")
```

```python
gpa = 4.4
is_active_in_student_council = True
has_volunteer_experience = True
scholarship_type = ""

if gpa >= 4.5 and is_active_in_student_council:
    scholarship_type = "full"
elif gpa >= 4.0 or has_volunteer_experience:
    scholarship_type = "partial"
elif not (gpa >= 4.5 and is_active_in_student_council) and not (gpa >= 4.0 or has_volunteer_experience):
    scholarship_type = "no"

print(scholarship_type)
```

```python
var_start = 1
var_end = 20
total_result = 0

for i in range(var_start, var_end + 1):
    if i % 3 == 0 or i % 5 == 0:
        continue
    if i % 2 == 0:
        total_result += i ** 2
    else:
        total_result -= i
print(total_result)
```

```python
var_start = 8
var_end = 20

for i in range(var_start, var_end + 1):
    if i % 7 == 0:
        print(f"Число {i}, яке ділиться на 7, знайдено. Завершення циклу.")
        break
else:
    print("Цикл завершено без знаходження числа яке ділиться на 7.")
```

```python
start_number = 15
end_number = 26

# i = start_number

while start_number <= end_number:
    if start_number % 2 != 0:
        start_number += 1
        continue
    if start_number % 7 == 0 and start_number % 3 != 0:
        print(f"Число {start_number}, є Щасливим числом.")
        break
    start_number += 1
else:
    print("No lucky number was found in the range.")
```

```python
raw_cars = "aUdi; bmW; pOrScHe ; toYoTa; MeRcEdEs "
cars_list = raw_cars.split(";")

for i in range(len(cars_list)):
    cars_list[i] = cars_list[i].strip()
    if cars_list[i].lower() == "bmw":
        cars_list[i] = cars_list[i].upper()
    else:
        cars_list[i] = cars_list[i].capitalize()

print(cars_list)

cars_list = cars_list[1::2]
final_string = "//".join(cars_list)
print(final_string)
```
