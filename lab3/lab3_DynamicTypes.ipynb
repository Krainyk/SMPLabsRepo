```python
inventory = [
    "Laptop:15",
    "Monitor:25",
    "Keyboard:50",
    "Mouse:45",
    "Webcam:10",
    "Headphones:30",
    "Laptop:5"
]

processed_inventory = []
for item in inventory:
    processed_item = item.split(":")
    processed_inventory.append(processed_item[0])

print(f"Список назв товарів (нормалізація): {processed_inventory}")

processed_inventory.append("Speaker")
processed_inventory.extend(["Printer", "Scanner"])
print(processed_inventory)

item_count = processed_inventory.count("Laptop")
print(f"Кількисть зходжень 'Laptop': {item_count}")

processed_inventory.sort()
print(f"Відсортований список: {processed_inventory}")

processed_inventory.remove("Mouse")
print(f"Список без Mouse: {processed_inventory}")

del processed_inventory[8]
print(f"Список без 9 елемента {processed_inventory}")
```

```python
transactions = [
{"id": "T001", "item": "Laptop", "quantity": 1, "price": 1200.00, "category":
"Electronics", "customer_id": "C101"},
{"id": "T002", "item": "Mouse", "quantity": 3, "price": 25.50, "category":
"Accessories", "customer_id": "C102"},
{"id": "T003", "item": "Keyboard", "quantity": 2, "price": 75.00, "category":
"Accessories", "customer_id": "C101"},
{"id": "T004", "item": "Laptop", "quantity": 1, "price": 1200.00, "category":
"Electronics", "customer_id": "C103"},
{"id": "T005", "item": "Webcam", "quantity": 5, "price": 40.00, "category":
"Accessories", "customer_id": "C102"}
]

new_transaction = {
    "id": "T006",
    "item": "Headphones",
    "quantity": 2,
    "price": 50.00,
    "category": "Accessories",
    "customer_id": "C104"
}

transactions.append(new_transaction)
print(f"Список транзакцій після додавання нового словника: {transactions}")

for t in transactions:
    if t["id"] == "T003":
        t["price"] = 70.00
        break
print(f"Список транзакцій після оновлення ціни для T003: {transactions}")

unique_customers = set()
for t in transactions:
    unique_customers.add(t["customer_id"])
print(f"Унікальні ідентифікатори клієнтів: {unique_customers}")

item_prices = []
for t in transactions:
    item_prices.append((t["item"], t["price"]))
print(f"Список товарів та їхніх цін: {item_prices}")

total = 0
for t in transactions:
    if t["category"] == "Electronics":
        total += t["quantity"]
print(f"Кількість товарів які були продані у категорії 'Electronics': {total}")


for i, t in enumerate(transactions):
    if t["id"] == "T005":
        transactions.pop(i)
        break
print(f"Список транзакцій після видалення словника з id 'T005): {transactions}")

sorted_by_price = sorted(transactions, key=lambda x: x["price"])
print("Дві найдешевші транзакції:")
for t in sorted_by_price[:2]:
    print(f"id: {t['id']}, item: {t['item']}, price: {t['price']}")
```

```python
projects = {
"Web_Redesign": ["Develop_Wireframes", "Design_Mockups",
"Implement_Frontend", "Test_UX"],
"Data_Migration": ["Extract_Legacy_Data", "Clean_Data", "Transform_Data",
"Load_New_DB"],
"Marketing_Campaign": ["Define_Audience", "Create_Content",
"Schedule_Posts", "Analyze_Metrics"]
}

projects["Mobile_App"] = ["Setup_Environment", "Develop_Backend"]
print(projects)

projects["Web_Redesign"].append("Deploy_Website")
print(projects)

projects["Data_Migration"].insert(0, "Backup_Original_Data")
print(projects)

# 2.1
all_tasks_flat = []
for tasks in projects.values():
    all_tasks_flat.extend(tasks)
print(f"Список всіх завдань (плоский список): {all_tasks_flat}")


# 2.2
unique_tasks = set(all_tasks_flat)
if len(all_tasks_flat) != len(unique_tasks):
    print("Є дублікати завдань між проєктами")
else:
    print("Дублікатів завдань немає")

# 3.1
removed_task = projects["Marketing_Campaign"].pop()
print(f"Видалене завдання: {removed_task}")

# 3.2
projects["Data_Migration"].remove("Clean_Data")

# 3.3
projects.pop("Mobile_App")

# 4.1
project_sizes = []
for name, tasks in projects.items():
    project_sizes.append((len(tasks), name))
sorted_projects = sorted(project_sizes, key=lambda p: p[0], reverse=True)
print(sorted_projects)

# 4.2
web_tasks = projects["Web_Redesign"]
index = web_tasks.index("Test_UX")
web_tasks[index] = "Final_UX_Review"
print(projects)
```

```python
employees = {
"E101": {"name": "Alice Johnson", "department": "Marketing", "salary": 65000,
"skills": ["SEO", "Content_Creation"]},
"E102": {"name": "Bob Smith", "department": "IT", "salary": 85000, "skills":
["Python", "Databases", "Networking"]},
"E103": {"name": "Charlie Brown", "department": "Marketing", "salary":
70000, "skills": ["Social_Media", "Analytics"]},
"E104": {"name": "Diana Prince", "department": "IT", "salary": 92000, "skills":
["Cloud", "Python", "Security"]}
}

# 1.1
employees["E105"] = {
    "name": "Eve Adams",
    "department": "Finance",
    "salary": 75000,
    "skills": ["Budgeting", "Reporting"]
}

# 1.2
employees["E102"]["department"] = "Development"

# 1.3
employees["E104"]["salary"] += 5000

# 2.1
all_departments = set()
for emp in employees.values():
    all_departments.add(emp["department"])
print(all_departments)

# 2.2
core_skills = set()
for emp in employees.values():
    core_skills.update(emp["skills"])
print(core_skills)

# 3.1
employees.pop("E103")

# 3.2
high_earners = []
for emp in employees.values():
    if emp["salary"] > 80000:
        high_earners.append((emp["name"], emp["salary"]))
print(high_earners)

# 4.1
sorted_by_name = sorted(employees, key=lambda emp_id: employees[emp_id]["name"])
print(sorted_by_name)

# 4.2
for dept in all_departments:
    count = 0
    for emp in employees.values():
        if emp["department"] == dept:
            count += 1
    print(f"{dept}: {count}")
```
