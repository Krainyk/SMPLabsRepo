```python
inventory = [
    "Laptop:15",
    "Monitor:25",
    "Keyboard:50",
    "Mouse:45",
    "Webcam:10",
    "Headphones:30",
    "Laptop:5"
]

processed_inventory = []
for item in inventory:
    processed_item = item.split(":")
    processed_inventory.append(processed_item[0])

print(f"Список назв товарів (нормалізація): {processed_inventory}")

processed_inventory.append("Speaker")
processed_inventory.extend(["Printer", "Scanner"])
print(processed_inventory)

item_count = processed_inventory.count("Laptop")
print(f"Кількисть зходжень 'Laptop': {item_count}")

processed_inventory.sort()
print(f"Відсортований список: {processed_inventory}")

processed_inventory.remove("Mouse")
print(f"Список без Mouse: {processed_inventory}")

del processed_inventory[8]
print(f"Список без 9 елемента {processed_inventory}")
```
